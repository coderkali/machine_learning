# DAF-18 in pictures

## 📖 The story: Asha madam opens the sealed envelope

### Who is Asha madam, and what must she decide?

Asha madam is the coordinator of a school in Delhi. Delhi's air is sometimes very bad. The tiny dust in the air is measured as **PM2.5**. A higher number means worse air, and **91 or more is "Poor"**. Children should not stand outside on a Poor day.

Every evening at 6 pm she must decide one thing:

> **Tomorrow morning, will the 600 children stand outside for assembly, or stay indoors?**

To decide, she needs a guess of **tomorrow's PM2.5**.

### Two ways she can guess

| | Way 1: her old method | Way 2: her model |
|---|---|---|
| Name | **Persistence** | **Ridge** (a small machine-learning model) |
| How it guesses | She looks at today's 5 pm reading and says, *"Tomorrow will be about the same."* | The computer looks at today's reading and the average of the last 3, 7 and 14 days, and predicts tomorrow's number |
| Cost | Free, no computer | Needs training, a file to save, a service to run |

The whole project asks one question: **is Way 2 really better than Way 1?** If it is not, why keep the computer?

### What she has done so far (a school comparison)

| Ticket | What happened | Like in school |
|---|---|---|
| DAF-14 to DAF-16 | She built the model and tried it | Learning the chapters and writing practice tests |
| DAF-17 | She tried the settings (the "knobs") using only the 251 **study days**. Then she wrote her final choice in `models/final_config.json` and **signed** it | Finishing preparation, writing the final study plan on paper, and signing it |
| **DAF-18 (now)** | She opens the **130 test days** (from 1 March 2026) | **The real board exam paper** |

Those 130 test days are the **sealed envelope**. The model has never been allowed to learn from them. That is why they can give an honest answer.

### The problem

The principal asks three plain questions:

> *"Does your model beat your old method? By how much? And on which days should I not trust it?"*

Asha madam is nervous. In the mock tests (DAF-17), her model won only a little: an average mistake of 36.5 against persistence's 37.9. That is a win of only about 1.4 out of about 37, which is far below the 10% the project asked for.

Now the envelope is on the table, and she feels a temptation:

```text
1. She opens the envelope and sees a poor score.
2. She quietly changes one setting.
3. She opens the envelope again. The score looks better.
4. She tells the principal the second score.
```

### What Sharma sir says

Sharma sir, the old maths teacher, stops her.

> *"Beta, the paper opens once. If the marks are poor, write the poor marks in the report card. A student who retakes the board paper until the marks look good has not passed. She has only learned the paper."*

**Why he is right.** The first score is the only honest one. The moment you change something because of what the test showed you, the test days have started to teach your model. They are no longer a fair test, like a student who has seen the question paper before the exam.

### So what do we do in this ticket?

```text
Open the envelope ONCE
   -> compare the model with persistence on the SAME 130 days
   -> say "met" or "not met" for each of the two pass marks
   -> open the answer sheet: which 10 days did we lose the most marks on, and why?
   -> write a one-page model card so that anyone can see where NOT to trust it
```

### One honest note

The test days were already looked at in earlier tickets (DAF-06 to DAF-16). So the envelope was not perfectly sealed. This is written down in decision D-007. It is why this notebook calls DAF-18 **the last look** at the test days, **not the first**, and why the model card must say it too.

## 🎯 The exact problem

> **How do we score the frozen model once, say honestly whether it met the goal, find the days where it fails, and write that down so that anyone who uses the model knows its limits?**

The answer is four things: **one test run**, **a verdict** against the two success rules, **the ten worst days**, and **a model card**.

The pictures for this story are below. This notebook goes with `18_final_test_and_model_card.ipynb`.

1. We have a **guess made by a model** and a **guess made by a simple free method**.
2. We kept some days **locked away** so we have an honest test.
3. We wrote the **pass marks first**.
4. We tested **once** and told the truth.
5. We looked at **where it fails**.
6. We saved it and wrote **instructions** for the next person.

> *"Ramesh bhai sells samosas. Every night he guesses tomorrow's sales. His old method is: same as today. He built a small model from the last year of sales. 1. Why did we lock some days away?
2. Why were the pass marks written **before** the test?
3. Why is the old method scored on the **same days**?
4. What does it mean if the model wins on MAE but misses more big days?
5. Why do we write down the **worst** days, not just the average?

✏️ If they answer these in their own words, they understood the ticket.

### 🎬 How we got here

**Sharma sir:** *"Beta, before you open it, tell me. What exactly will you do?"*

### 😖 The problem

Two easy mistakes can spoil this step, and neither will show an error message:

| Mistake | What it looks like | Why it spoils the result |
|---|---|---|
| Giving the two students different papers | Ridge scored on 130 rows, persistence on 161 rows (its old DAF-06 number) | One student got an easier paper. The comparison is a joke |
| Opening the envelope twice | "The first score looked odd, let me change one thing and run it again" | The test days start teaching the model. This is the trap in the ticket |

### 💡 The fix, in four small moves

```text
1. COOK     rebuild Ridge from final_config.json, fit it on all 251 study days
2. SERVE    predict tomorrow's PM2.5 for each of the 130 test days
3. OLD WAY  persistence forecast = the day's pm25_until_17, nothing learned
4. MARK     put both forecasts next to the real value and calculate the marks
```

### What each move means

| Move | In school words | What happens in the code |
|---|---|---|
| 1 · Cook | She studies all her chapters one last time before the exam | `StandardScaler` and `Ridge(alpha=1.0)` are fitted on the 251 train rows. The scaler learns its average and spread from train rows only |
| 2 · Serve | She writes the answers on the board paper | The model gets the 4 features of each test day and gives one forecast per day. It does not learn anything here |
| 3 · Old way | The old teacher just copies today's number | The persistence forecast is the feature `pm25_until_17` itself. There is nothing to fit |
| 4 · Mark | The teacher checks the answer sheet | Two marks per student: MAE and recall (below) |

### The two marks, in simple words

**MAE (mean absolute error), the average mistake.**
If the real value is 100 and the forecast is 80, the mistake is 20. If the real value is 60 and the forecast is 70, the mistake is 10. We ignore the plus or minus sign, and take the average over 130 days. **Smaller is better.** It is in µg/m³, so it is easy to explain: *"On an average day we are off by this much."*

**Recall on Poor days, the smoggy days she did not miss.**
In the 130 test days, 17 were really Poor (PM2.5 of 91 or more). We count how many of those 17 the student also called Poor. If she caught 12 of 17, recall is 12 ÷ 17 = 71%. **Higher is better.** A missed Poor day means children stood in smog, so this is the safety mark.

### Look at

- **Black line:** what really happened on each of the 130 days
- **Grey line:** persistence. It tends to run one day late, like a student who copies yesterday's answer
- **Blue line:** Ridge
- **Red dashed line at 91:** the Poor line. The days where the black line is above it are the ones Asha cares about most
- **Small bars on the right:** the MAE of each, on **the same 130 rows**

### Questions to ask when the real picture arrives

1. Is the blue line closer to the black one, or only a little closer?
2. At the peaks (the black spikes above 91), does blue rise in time, or does it also come one day late?
3. Do both lines miss the same Poor days, or different ones?

### The rule that cannot be broken in this step

**The test rows are predicted in exactly one cell, exactly one time.** If the numbers look odd, we do **not** rerun with a change. We write them down and move to Step 4. Anything odd is for the error analysis in Step 6 to explain.

| 5 Verdict | Compare the marks with the pass marks. Say MET or NOT MET honestly |

| 7 Save + reload | Save the model to a file, load it back, and check it gives the same answer |
| 8 Model card | One page: what it does, how good it is, when **not** to trust it |
| 9 Log | One line in `experiments.csv`, so the result is never lost |

### What each box does, in more detail

Read every box with the same five questions:

- **Goes in:** what the box needs before it can start
- **What happens:** the work done inside the box
- **Comes out:** what it hands to the next box
- **If we skip it:** what goes wrong
- **In the toy:** the real number from the toy town below

---

#### Box 1 · Rebuild

| | |
|---|---|
| **Goes in** | The signed recipe from DAF-17 (model = Ridge, `alpha = 1.0`, which features), and the study rows |
| **What happens** | First, a check that the study data is the same data we signed with. We add up the answers and compare the total (the "fingerprint"). Then we train Ridge again on the **study rows only** |
| **Comes out** | A trained model, ready to make guesses |
| **If we skip it** | We might test a model that is different from the one we signed, or one trained on changed data. Then the test is about the wrong model |
| **In the toy** | Fingerprint = 1315, same as when signed ✔. The model learnt: tomorrow = 0.30 × today + 65 |

#### Box 2 · Pass marks

| | |
|---|---|
| **Goes in** | The project's success rules from `docs/requirements.md` |
| **What happens** | We write the two rules down **before** any test result exists. Rule 1: the average mistake is at least 10% smaller than the old method's. Rule 2: catch at least as many Poor days as the old method |
| **Comes out** | Two empty boxes, waiting for the results |
| **If we skip it** | After seeing the result, it is very tempting to say "8% is good enough". Rules written after the result always seem to bend toward the result |
| **In the toy** | 10% and "at least as many", written before Box 3 |

#### Box 3 · Open envelope (the only box that touches the sealed rows)

| | |
|---|---|
| **Goes in** | The trained model from Box 1, and the sealed rows |
| **What happens** | For every sealed row, we make two guesses: Ridge's guess, and the old method's guess (just today's number). Then we compare both with what really happened and calculate the marks: the average mistake (MAE), and how many Poor days each one caught |
| **Comes out** | One table of guesses and mistakes. **Every later box reads this table. Nothing touches the sealed rows again** |
| **If we skip it, or do it twice** | Without it, we have no honest result. If we do it twice ("let me change one setting and try again"), the sealed rows start teaching the model, and the test stops being honest |
| **In the toy** | Average mistake: old method 18.6, Ridge 15.0 |

#### Box 4 · Second opinion

| | |
|---|---|
| **Goes in** | The study rows only, plus the same recipe |
| **What happens** | A few small exams inside the study rows. Each exam learns from the earlier rows and checks on the next few rows (walk-forward). Then we see whether Ridge wins there too |
| **Comes out** | A small table: who won in each small exam. In the real notebook these numbers are not run again; they are copied from DAF-16 |
| **If we skip it** | With only one test, a win (or a loss) might be luck. The sealed rows might just happen to be easy or hard days |
| **In the toy** | Ridge wins both small exams: 21.0 vs 21.3 (only just), and 8.9 vs 14.0 |

#### Box 5 · Verdict

| | |
|---|---|
| **Goes in** | The marks from Box 3, and the pass marks from Box 2 |
| **What happens** | We compare them and write **MET** or **NOT MET** for each rule. Both rules must be met |
| **Comes out** | One honest sentence: "the model is / is not better than the free method" |
| **If we skip it** | People see numbers but not the answer. Or someone reads them in the way that suits them |
| **In the toy** | Rule 1: 19% ≥ 10% ✔. Rule 2: 3 of 4 = 3 of 4 ✔. **MET** |

#### Box 6 · Worst days

| | |
|---|---|
| **Goes in** | The table from Box 3 |
| **What happens** | Sort the rows by Ridge's mistake, biggest first. For the top ones, look at what happened that day and write a likely reason |
| **Comes out** | A short list of failure **patterns** (for example, "sudden changes fool it") |
| **If we skip it** | The average hides the dangerous days. A model that is good on average can still be badly wrong on exactly the days that matter |
| **In the toy** | Row 20: sudden clean-up (mistake 27). Row 15: smog stayed high (mistake 22) |

#### Box 7 · Save + reload

| | |
|---|---|
| **Goes in** | The trained model from Box 1 |
| **What happens** | Save it to a file with `joblib.dump`, together with the feature names and the study dates. Start fresh, load the file with `joblib.load`, and predict one row again |
| **Comes out** | A model file that we **know** gives the same answer |
| **If we skip it** | The live service (DAF-19/20) loads the file. If the file is wrong, the service runs a different model from the one we tested, and nobody notices |
| **In the toy** | Row 15: 100.7 before saving, 100.7 after loading ✔ |

#### Box 8 · Model card

| | |
|---|---|
| **Goes in** | Everything above: the recipe, the data, the marks, the verdict, the worst days |
| **What happens** | Write one page: what the model is for, which data it learnt from, its features, its marks, the Poor threshold (91), **known failure cases**, and what it must **not** be used for |
| **Comes out** | `docs/model_card.md`, a page a stranger can read in two minutes |
| **If we skip it** | Someone uses the model somewhere it was never tested (another station, another city) and trusts it blindly |
| **In the toy** | Fails on sudden swings. Missed Poor row 21. Not for other towns |

#### Box 9 · Log

| | |
|---|---|
| **Goes in** | The final numbers from Box 3 and the verdict from Box 5 |
| **What happens** | Add one row to `reports/experiments.csv` |
| **Comes out** | A permanent record in `reports/experiments.csv`, next to every earlier experiment |
| **If we skip it** | In three months nobody remembers what the final test said, or which model it was |
| **In the toy** | `toy, Ridge, MAE 15.0, old 18.6, Poor 3 of 4, MET` |

### What each box hands to the next

```text
 recipe + study rows ─► 1 trained model
                        2 pass marks (written, still empty)
 model + sealed rows ─► 3 table of guesses and mistakes   ◄── the ONLY time the sealed rows are used
 study rows only     ─► 4 small exams: does Ridge win there too?
 table (3) + marks (2) ► 5 MET / NOT MET
 table (3)           ─► 6 worst rows ─► failure patterns
 model (1)           ─► 7 model file, checked
 all of the above    ─► 8 model card
 numbers (3) + (5)   ─► 9 one row in experiments.csv
```

✏️ Boxes 1 to 3 **produce** the result. Boxes 4 to 9 only **read** it, check it, explain it and save it. None of them is allowed to change the model.

## Step 1 · Rebuild the model, and see what each method believes

![What each method believes](figures/18_walkthrough/toy/toy_03_learn.png)

**Asha:** *"Sir, I cook the dish again from my signed recipe: Ridge, `alpha = 1.0`, one feature `today`. And I use only the 14 study rows."*

**Sharma sir:** *"And how do you know nobody changed your study book?"*

**Asha:** *"I add up the 14 answers. The total is **1315**, the same as when I signed. That is my fingerprint."*

### How to read the picture

| Line | What it means |
|---|---|
| **Grey line** (old method) | "Tomorrow = today." Whatever today is, it says the same number again |
| **Blue line** (Ridge) | What Ridge learnt from the 14 dots: **tomorrow = 0.30 × today + 65** |
| **Black dots** | The 14 study rows. Each dot is one real (today, tomorrow) pair |

**The one idea in this picture.** The blue line is flatter than the grey one. In this town, a very high day is usually followed by a lower one, and a very low day by a higher one. The old method cannot know that. Ridge learnt it from the dots.

✏️ The sealed rows are **not** in this picture. The model has never seen them.

## Step 2 · Write the pass marks before opening the envelope

![Pass marks](figures/18_walkthrough/toy/toy_02_pass_marks.png)

**Sharma sir:** *"Write the pass marks on the board now. If you write them after you see the marks, you will choose the pass marks that suit you."*

| Rule | Pass mark | Why this rule exists |
|---|---|---|
| 1 · Average mistake (MAE) | Ridge's mistake is at least **10% smaller** | A tiny win is not worth the extra work of running a model |
| 2 · Catching Poor days | Ridge catches **at least as many** Poor days | A model that is more accurate on average but misses smoggy days would put children outside on bad air |

Both rules must hold. The two boxes stay empty (**?**) until Step 3.

## Step 3 · Open the envelope, once

![Open the envelope](figures/18_walkthrough/toy/toy_04_open_envelope.png)

**Sharma sir:** *"Same 7 rows for both methods. One time only. Then close the envelope forever."*

### How to read the picture

- The **black dash** is what really happened.
- The **grey square** is the old method's guess. The **blue dot** is Ridge's guess.
- The **vertical line** from the dash to the guess is the **mistake**. The number next to it is its size.

### The sums

| row | real | old method says | Ridge says | old mistake | Ridge mistake |
|---:|---:|---:|---:|---:|---:|
| 15 | 123 | 116 | 100.7 | 7 | 22.3 |
| 16 | 109 | 123 | 102.8 | 14 | 6.2 |
| 17 | 91 | 109 | 98.5 | 18 | 7.5 |
| 18 | 73 | 91 | 93.1 | 18 | 20.1 |
| 19 | 85 | 73 | 87.6 | 12 | 2.6 |
| 20 | 64 | 85 | 91.2 | 21 | 27.2 |
| 21 | 104 | 64 | 84.8 | 40 | 19.2 |
| | | | **average = MAE** | **18.6** | **15.0** |

**MAE** is just "add the mistakes, divide by 7". The plus or minus sign is ignored. On an average sealed day, the old method was off by 18.6 and Ridge by 15.0.

✏️ Ridge does **not** win every row. The old method wins on rows 15 and 18. MAE looks at the average, not at single rows.

## Step 4 · Second opinion: do smaller tests agree?

![Second opinion](figures/18_walkthrough/toy/toy_05_second_opinion.png)

**Sharma sir:** *"One exam can be luck. If the monthly tests say the same thing, then it is real."*

**What we do.** Inside the **study rows** only, we run two small exams. Each one learns from the past and is checked on the next few rows. This is called **walk-forward**. The sealed rows are not used.

| Exam | Learn from | Checked on | Old method MAE | Ridge MAE |
|---|---|---|---:|---:|
| Fold 1 | rows 1-7 | rows 8-10 | 21.3 | 21.0 |
| Fold 2 | rows 1-10 | rows 11-14 | 14.0 | 8.9 |
| Main test (Step 3) | rows 1-14 | rows 15-21 | 18.6 | 15.0 |

**Read it like this:** Ridge is lower in all three. In fold 1 it only just wins. So the win is not a lucky accident of the 7 sealed rows. It shows up every time, sometimes small, sometimes big.

✏️ If the second opinion had disagreed, we would **not** change anything. We would write the disagreement in the model card.

## Step 5 · The verdict

![Verdict](figures/18_walkthrough/toy/toy_06_verdict.png)

**Sharma sir:** *"Now read the marks aloud, against the pass marks you wrote in Step 2."*

| Rule | Pass mark | Result | |
|---|---|---|---|
| 1 · MAE | at least 10% smaller | (18.6 − 15.0) ÷ 18.6 = **19% smaller** | ✔ MET |
| 2 · Poor days | catch at least as many | old method **3 of 4**, Ridge **3 of 4** | ✔ MET |

**Both met, so Ridge is worth keeping** in the toy town.

Look at row 21 in the picture: real 104, but **both** methods said less than 91. Both missed it. Rule 2 only asks "not worse than the old method", so it still passes. But a missed Poor day is a real danger, and it must go into the model card.

✏️ In the real notebook the verdict may be **NOT MET**. That is a fine and honest result. The job of this step is to tell the truth, not to pass.

## Step 6 · Where did Ridge lose the most marks?

![Worst days](figures/18_walkthrough/toy/toy_07_worst_days.png)

**Sharma sir:** *"Don't only count the marks. Open the answer sheet. Which questions did you lose, and why?"*

| Rank | Row | today | real tomorrow | Ridge said | Mistake | Likely reason |
|---:|---:|---:|---:|---:|---:|---|
| 1 | 20 | 85 | 64 | 91.2 | 27 | The air suddenly cleaned up overnight. Nothing in "today" warned of it |
| 2 | 15 | 116 | 123 | 100.7 | 22 | The smog stayed high, but Ridge always pulls its guess toward the middle (about 94) |

**Why this matters.** These are not random. They are **patterns**:

- Sudden swings fool it (rows 20 and 21).
- In a long smog spell it guesses too low (row 15).

Patterns go into the model card as **known failure cases**. In the real notebook this is the ten worst days, not two.

## ✅ Check it yourself: the toy in plain code

This cell does Steps 1, 3, 5 and 6 on the toy data. The printed numbers match the pictures above. (The pictures are drawn by `figures/18_walkthrough/make_toy_figures.py`.)

````python
# The whole toy, steps 1 to 7, in plain pandas + sklearn. Made-up numbers, no project files.
import pandas as pd
from sklearn.linear_model import Ridge

readings = [115, 103, 113, 71, 69, 83, 87, 108, 120, 110, 68,
            85, 93, 89, 116, 123, 109, 91, 73, 85, 64, 104]
toy = pd.DataFrame({"row": range(1, 22), "today": readings[:-1], "tomorrow": readings[1:]})
study, sealed = toy[toy.row <= 14], toy[toy.row >= 15].copy()

# Step 1: rebuild from the recipe, learn from study rows only, check the fingerprint
model = Ridge(alpha=1.0).fit(study[["today"]], study["tomorrow"])
print(f"Step 1  rule: tomorrow = {model.coef_[0]:.2f} x today + {model.intercept_:.1f}"
      f"   fingerprint = {study['tomorrow'].sum()}")

# Step 3: open the envelope ONCE
sealed["old_method"] = sealed["today"]
sealed["ridge"] = model.predict(sealed[["today"]]).round(1)
mae_old = (sealed["old_method"] - sealed["tomorrow"]).abs().mean()
mae_ridge = (sealed["ridge"] - sealed["tomorrow"]).abs().mean()
print(sealed.to_string(index=False))

# Step 5: verdict against the pass marks written in Step 2
improve = (mae_old - mae_ridge) / mae_old
poor = sealed[sealed["tomorrow"] >= 91]
caught_old, caught_ridge = (poor["old_method"] >= 91).sum(), (poor["ridge"] >= 91).sum()
print(f"Step 5  MAE {mae_old:.1f} -> {mae_ridge:.1f} = {improve:.0%} better  ->  rule 1 {'MET' if improve >= 0.10 else 'NOT MET'}")
print(f"        Poor days caught: old {caught_old} of {len(poor)}, Ridge {caught_ridge} of {len(poor)}"
      f"  ->  rule 2 {'MET' if caught_ridge >= caught_old else 'NOT MET'}")

# Step 6: worst rows for Ridge
sealed["mistake"] = (sealed["ridge"] - sealed["tomorrow"]).abs()
print("Step 6  the two worst rows:")
print(sealed.nlargest(2, "mistake")[["row", "today", "tomorrow", "ridge", "mistake"]].to_string(index=False))
````
