# DAF-16 in pictures

No code. Almost no text. One tiny example, one picture per step. Read the pictures in order.

- Picture 1 is the problem. Pictures 2 to 10 are Steps 1 to 8 of `16_walk_forward.ipynb`.
- Pictures 2, 3, 4 use real data or the 8-day example from Step 1. Pictures 6 and 7 use tiny made-up numbers to show how to read the results. The real numbers come when we run Step 4.
