```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
tips = px.data.tips()
tips
```

```python
df = px.data.gapminder().query("country=='India'")
df.head()
```

```python
fig = px.line(df, x="year",y="lifeExp")
fig.show()
```

```python
fig = px.line(
 df,
 x="year",
 y="lifeExp",
 markers=True
)
fig.show()
```

```python
fig = px.line(
 df,
 x="year",
 y="lifeExp",
 title="Life Expectancy of India"
)
fig.show()
```

```python
fig = px.scatter(
 tips,
 x="total_bill",
 y="tip"
)
fig.show()
```

```python
fig = px.scatter(
 tips,
 x="total_bill",
 y="tip",
 color="sex"
)
fig.show()
```

```python
fig = px.scatter(
 tips,
 x="total_bill",
 y="tip",
 size="size"
)
fig.show()
```

```python
fig = px.bar(
 tips,
 x="day",
 y="total_bill"
)
fig.show()
```

```python
fig = px.bar(
 tips,
 x="day",
 y="total_bill",
 color="sex"
)
fig.show()
```

```python
fig = px.bar(
 tips,
 x="total_bill",
 y="day",
 orientation="h"
)
fig.show()
```
