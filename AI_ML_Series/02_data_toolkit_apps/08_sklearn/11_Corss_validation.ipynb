# Cross Validation

> 📘 **Instructor Curriculum** — M07 Machine Learning, scikit-learn module.

Every notebook so far ended the same way: split once, fit, `.score()`, read one number.
That number depends on which rows happened to land in the test set.

Cross validation replaces the single split with **many splits**. It scores each one and
reports the average.

```text
   ONE SPLIT                        CROSS VALIDATION (5 rounds)

     data                                    data
       |                                       |
   split once                            split 5 ways
       |                             ______/ / | \ \______
     T | t                          T|t  T|t  T|t  T|t  T|t
       |                             |    |    |    |    |
     model                          mdl  mdl  mdl  mdl  mdl
       |                             |    |    |    |    |
    1 score                          s1   s2   s3   s4   s5
                                      \____\___|___/____/
                                               |
                                            average

   T = train rows      t = test rows
```

The data is **Iris**: 150 rows, 4 columns, 3 classes of 50 rows each, and it is already
**sorted by class**. That last fact causes everything interesting in this notebook.

```text
   row    0 ....................  49    class 0   setosa
   row   50 ....................  99    class 1   versicolor
   row  100 .................... 149    class 2   virginica
```

The model is `GaussianNB()` in every section. Only the **splitting strategy** changes.

````python
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
table = pd.DataFrame(iris.data,columns=iris.feature_names)
table.head()
````

````python
iris
````

````python
table['target']= iris.target
table.head()
````

````python
x = table.drop(['target'], axis='columns')
y = iris.target
````

## 1. The baseline — one `train_test_split`

120 rows to learn from, 30 rows kept hidden. `random_state=0` fixes which rows go where.

`classifier1.score(X_test, y_test)` → **`0.966667`**, so 29 of the 30 hidden rows were
correct.

That is a real number, but it is **one number from one draw**. Change `random_state` and it
moves. Everything below is about not trusting a single draw.

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.20, random_state = 0)
````

````python
from sklearn.naive_bayes import GaussianNB
classifier1 = GaussianNB()
classifier1.fit(X_train, y_train)
````

````python
classifier1.score(X_test,y_test)
````

## 2. `KFold(n_splits=5)`

KFold cuts the 150 rows into 5 blocks of 30 and runs 5 rounds. Each round uses one block as
test and the other four as train, so **every row is tested exactly once**.

```text
   150 rows / 5 folds = 30 rows per fold

   rows   0 ...... 29 | 30 ..... 59 | 60 ..... 89 | 90 .... 119 | 120 ... 149
            fold 1        fold 2        fold 3        fold 4        fold 5

   round 1   [ t ][ T ][ T ][ T ][ T ]    test 0-29      train 120 rows
   round 2   [ T ][ t ][ T ][ T ][ T ]    test 30-59     train 120 rows
   round 3   [ T ][ T ][ t ][ T ][ T ]    test 60-89     train 120 rows
   round 4   [ T ][ T ][ T ][ t ][ T ]    test 90-119    train 120 rows
   round 5   [ T ][ T ][ T ][ T ][ t ]    test 120-149   train 120 rows
```

The `TRAIN-->` / `TEST-->` arrays printed below are exactly those index blocks.
`shuffle=False` is the default, so the blocks are taken in the order the rows already sit in.

`cross_val_score(classifier1, x, y, cv=kf)` fits a **fresh copy** of the model on every
round. The earlier `classifier1.fit(X_train, y_train)` is not reused.

| round | test rows | score |
|---|---|---:|
| 1 | 0-29 | `1.000000` |
| 2 | 30-59 | `0.966667` |
| 3 | 60-89 | `0.900000` |
| 4 | 90-119 | `0.933333` |
| 5 | 120-149 | `0.933333` |
| | **mean** | **`0.946667`** |

### The problem hiding inside those folds

Iris is sorted by class, so counting what each fold actually holds:

| fold | rows | class 0 | class 1 | class 2 |
|---|---|---:|---:|---:|
| 1 | 0-29 | 30 | 0 | 0 |
| 2 | 30-59 | 20 | 10 | 0 |
| 3 | 60-89 | 0 | 30 | 0 |
| 4 | 90-119 | 0 | 10 | 20 |
| 5 | 120-149 | 0 | 0 | 30 |

**No fold ever contains all three classes.** Round 1 tests only setosa, and trains on a set
holding just 20 setosa rows out of 120. Each round is answering an easier or stranger
question than the real problem. That is what the next splitter fixes.

````python
from sklearn.model_selection import cross_val_score,KFold
````

````python
kf=KFold(n_splits=5)
kf
````

````python
for train, test in kf.split(x, y):
    print("TRAIN-->", train,'\n', "TEST-->", test,'\n')
````

````python
score=cross_val_score(classifier1,x,y,cv=kf)
````

````python
score
````

````python
score.mean()
````

## 3. `StratifiedKFold(n_splits=5)`

**Stratified** means: keep the class mix of the full data inside every fold. Iris is
50 / 50 / 50, so each fold of 30 must be 10 / 10 / 10.

```text
   class 0    row   0 ................  49
   class 1    row  50 ................  99
   class 2    row 100 ................ 149

                 take 10 from each block, every round

   fold 1 test =    0- 9  +   50- 59  +  100-109     10 + 10 + 10 = 30
   fold 2 test =   10-19  +   60- 69  +  110-119
   fold 3 test =   20-29  +   70- 79  +  120-129
   fold 4 test =   30-39  +   80- 89  +  130-139
   fold 5 test =   40-49  +   90- 99  +  140-149
```

The first `TEST-->` array printed below is `[0..9, 50..59, 100..109]`. The diagram is
literally the output.

`skf.get_n_splits(x, y)` returns `5` — the number of **rounds**, not the fold size.

| round | score |
|---|---:|
| 1 | `0.933333` |
| 2 | `0.966667` |
| 3 | `0.933333` |
| 4 | `0.933333` |
| 5 | `1.000000` |
| **mean** | **`0.953333`** |

Plain KFold gave `0.946667`, stratified gives `0.953333`. The gap is small here because
GaussianNB is strong on Iris; on an imbalanced dataset a plain KFold fold can end up with
**zero rows** of a rare class and the score stops meaning anything.

For a classifier, `cross_val_score(model, x, y, cv=5)` already uses `StratifiedKFold` behind
the scenes. Passing a plain `KFold` is the choice that needs a reason.

````python
from sklearn.model_selection import StratifiedKFold
````

````python
skf = StratifiedKFold(n_splits=5)
skf
````

````python
skf.get_n_splits(x, y)
````

````python
for train, test in skf.split(x, y):
    print("TRAIN-->", train,'\n', "TEST-->", test,'\n')
````

````python
classifier2 = GaussianNB()
````

````python
score=cross_val_score(classifier2,x,y,cv=skf)
````

````python
score
````

````python
score.mean()
````

## 4. `LeavePOut(p=2)`

No blocks this time. Take **every possible pair of rows** as the test set.

```text
   test = 2 rows          train = 148 rows

   round     1     test = (  0,   1)
   round     2     test = (  0,   2)
     ...
   round   149     test = (  0, 149)
   round   150     test = (  1,   2)
     ...
   last round      test = (148, 149)

   number of rounds = 150 x 149 / 2 = 11,175
```

`lp.get_n_splits(x)` prints `11175`, and `cross_val_score` fits GaussianNB **11,175 times**.

Each round is scored on 2 rows, so one score can only be `0.0`, `0.5` or `1.0`. The array is
almost all `1.` and no single entry is informative — only the mean is.

**mean = `0.953378`**

This is the most thorough option and the least usable one. 150 rows already costs 11,175
fits; 1,000 rows would cost 499,500.

````python
from sklearn.model_selection import LeavePOut
````

````python
lp=LeavePOut(p=2)
lp
````

````python
lp.get_n_splits(x)
````

````python
for train, test in lp.split(x, y):
    print("TRAIN-->", train,'\n', "TEST-->", test,'\n')
````

````python
classifier3 = GaussianNB()
````

````python
score=cross_val_score(classifier3,x,y,cv=lp)
````

````python
score
````

````python
score.mean()
````

## 5. `LeaveOneOut()`

The same idea with `p = 1`. One row is the test set, the other 149 train.

```text
   round     1     test = row   0     train = 149 rows
   round     2     test = row   1     train = 149 rows
     ...
   round   150     test = row 149     train = 149 rows

   number of rounds = 150 = one per row
```

Each score is `1.` (that row was predicted correctly) or `0.` (wrong), so the array is a
**per-row report card**, not a list of accuracies.

**mean = `0.953333` = 143 / 150**, so 7 rows were wrong. The array shows exactly which:

```text
   wrong rows:  52, 70, 77   (class 1, versicolor)
                106, 119, 133, 134   (class 2, virginica)

   setosa rows 0-49:  0 mistakes
```

Every mistake is versicolor confused with virginica — the same two overlapping classes the
pair plot showed in the EDA notebook. Setosa is never missed. A plain 5-fold mean of
`0.9533` cannot tell you that; LeaveOneOut can.

````python
from sklearn.model_selection import LeaveOneOut
````

````python
lo=LeaveOneOut()
````

````python
for train, test in lo.split(x, y):
    print("TRAIN-->", train,'\n', "TEST-->", test,'\n')
````

````python
classifier4 = GaussianNB()
````

````python
score=cross_val_score(classifier4,x,y,cv=lo)
````

````python
score
````

````python
score.mean()
````

## 6. Which one to use

| splitter | rounds | test size per round | mean score | cost |
|---|---:|---|---:|---|
| `train_test_split` | 1 | 30 | `0.966667` | 1 fit |
| `KFold(5)` | 5 | 30 (one class block) | `0.946667` | 5 fits |
| `StratifiedKFold(5)` | 5 | 30 (10 per class) | `0.953333` | 5 fits |
| `LeavePOut(p=2)` | 11,175 | 2 | `0.953378` | 11,175 fits |
| `LeaveOneOut()` | 150 | 1 | `0.953333` | 150 fits |

The single split reported `0.9667`. Three independent thorough methods all land on about
`0.9533`. That extra `0.013` was **luck in the draw**, not skill in the model.

```text
   how many rows?          which splitter?

   normal dataset    ->    StratifiedKFold(5)  or  cv=5      the default answer
   imbalanced        ->    StratifiedKFold                   never plain KFold
   very few rows     ->    LeaveOneOut                       150 fits is affordable
   research / tiny   ->    LeavePOut                         rarely worth the cost
```

**If you remember only one thing:** a score from one split is a sample of size one. Cross
validation is that same score measured many times and averaged, and on a classifier
`StratifiedKFold` is the version to reach for, because it keeps every class present in every
fold.
