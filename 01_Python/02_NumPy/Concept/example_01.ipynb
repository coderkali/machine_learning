# NumPy Arrays — Example 01

NumPy is the library that gives Python a real array type.

A Python list can hold anything — a string next to a number next to a boolean.
A NumPy array holds **one type only**, packed together in one block of memory.

Backend analogy: a Python list is like `ArrayList<Object>`. A NumPy array is
like `int[]`. The `int[]` is stricter, but it is much faster and it lets you do
math on the whole array in one line.

This notebook walks through:

1. A plain Python list (the starting point)
2. Making an array with `np.array()`
3. Dimensions — 0-D, 1-D, 2-D, 3-D, 4-D
4. Data types inside an array
5. Indexing (positive and negative)
6. Slicing
7. Looping and searching
8. Sorting, inserting, deleting, flipping, appending

## 1. Start with a plain Python list

Before NumPy, this is what we had. A list can mix types and it grows freely.

```python
arr1 = ['pune', "mumbai", 'surat', "kolkata", 'bihar', "jaipur", 'chennai']
arr1
```

```python
type(arr1)
```

```python
# lists are indexed from 0
arr1[0]
```

`type(arr1)` says `list`. Index `0` is the first item, not the second. That
part is the same in NumPy.

## 2. Turn a list into a NumPy array

`numpy.array()` takes a list and gives back an `ndarray` — the NumPy array
type. The name `ndarray` means "n-dimensional array".

```python
import numpy

arr2 = numpy.array([10, 20, 30, 40, 50])
arr2
```

```python
type(arr2)
```

The type is `numpy.ndarray`, not `list`. It looks like a list on screen but it
is a different object with different powers.

Now see the two ways of showing it.

```python
print(arr2)
```

```python
arr2
```

`print(arr2)` shows `[10 20 30 40 50]` — no commas, just spaces. That is the
friendly view.

Typing `arr2` alone shows `array([10, 20, 30, 40, 50])` — the exact code you
would write to rebuild it. That is the developer view. Same array, two ways of
printing.

A tuple works too. NumPy only needs something it can walk through.

```python
arr3 = numpy.array((100, 200, 300, 400, 500))
arr3
```

```python
type(arr3)
```

Still `numpy.ndarray`. The tuple was only the input container — once inside
NumPy it is an array like any other.

## 4. Dimensions — how many brackets deep is the data?

Dimension = how many index numbers you need to reach one value.

| Dimensions | Called | Real-world shape | Index needed |
|---|---|---|---|
| 0-D | scalar | a single number | none |
| 1-D | vector | a row of numbers | `a[i]` |
| 2-D | matrix | a table (rows × columns) | `a[i, j]` |
| 3-D | tensor | a stack of tables | `a[i, j, k]` |

`.ndim` tells you the number of dimensions. `.shape` tells you how big each
dimension is.

```python
# 0-D array (scalar)
a0 = np.array(5)
a0
```

```python
a0.ndim, a0.shape
```

`ndim` is `0` and `shape` is `()` — an empty tuple. There is no length to
report because there is no axis at all. It is one lonely number wrapped in a
NumPy box.

```python
# 1-D array
a1 = np.array([0, 1, 2, 3])
a1
```

```python
a1.ndim, a1.shape
```

One pair of brackets → `ndim` is `1`. Shape `(4,)` means one axis holding 4
items. The trailing comma is just Python's way of writing a one-item tuple.

```python
# 2-D array
a2 = np.array([[1, 2, 3], [4, 5, 6]])
a2
```

```python
a2.ndim, a2.shape
```

Brackets inside brackets → `ndim` is `2`. Shape `(2, 3)` reads as
**2 rows, 3 columns**. Rows always come first.

```python
# 3-D array
a3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
a3
```

```python
a3.ndim, a3.shape
```

Three levels of brackets → `ndim` is `3`. Shape `(2, 2, 2)` reads as
**2 blocks, each 2 rows, each row 2 columns**.

```python
# 4-D array -> each extra pair of brackets adds one more dimension
a4 = np.array([[[[4, 44, 444, 4444]]]])
a4
```

```python
a4.ndim, a4.shape
```

Shape `(1, 1, 1, 4)` — three wrapper layers of size 1, then the 4 real values.
The wrappers hold no extra data. They only add depth.

There is a shortcut for this. `ndmin=` tells NumPy the minimum number of
dimensions you want, and it adds the wrapper brackets for you.

```python
a4b = np.array([4, 44, 444, 4444], ndmin=4)
a4b
```

```python
a4b.ndim, a4b.shape
```

Same result as writing `[[[[...]]]]` by hand — `ndim` 4, shape `(1, 1, 1, 4)`.
Less typing and far fewer mismatched brackets.

## 5. Data types — one type for the whole array

This is the biggest difference from a Python list. A list keeps each item's own
type. An array must pick **one type for everything**.

If you hand it mixed types, NumPy does not throw an error. It quietly promotes
everything to the one type that can hold them all. This is called **upcasting**.

The promotion ladder is: `bool` → `int` → `float` → `string`. The widest type
in your data wins.

```python
amix1 = np.asarray([True, 5, 5.0, "Kali"])
amix1
```

A string was in the list, so **everything** became a string. Look at the
output: `'True'`, `'5'`, `'5.0'` are now text in quotes. The dtype `<U21` means
"unicode string, up to 21 characters".

Once this happens you cannot do maths on the array any more. This is a common
silent bug — one stray text value in a CSV column turns the whole column into
strings.

```python
amix1.dtype
```

Now the same idea with only booleans and numbers.

```python
amix2 = np.array([True, 5])
amix2
```

`True` became `1`. Booleans sit at the bottom of the ladder, so integers win.

```python
amix3 = np.array([False, 10])
amix3
```

`False` became `0`. Same rule.

```python
amix4 = np.array([True, 'shridhar', 5, 5.0])
amix4
```

A string is present again, so all four values became text — `dtype='<U32'`.

### Choosing the type yourself

You do not have to accept NumPy's guess. Pass `dtype=` and it converts for you.

```python
amix5 = np.asarray([False, 10], dtype=np.int32)
amix5
```

`int32` means each number uses 32 bits. NumPy's default on most machines is
`int64`. Choosing `int32` halves the memory when you know the values are small
— the same trade-off as picking `int` over `long` in Java.

```python
amix6 = np.array([1, 5.55, 2, 6.9], dtype=np.float64)
amix6
```

`1` and `2` are shown as `1.` and `2.` — the trailing dot means they are now
floats, not integers. The whole array is `float64`.

## 6. Indexing — reading one value

Give one index number per dimension. A 2-D array needs two numbers, a 3-D array
needs three, and so on. Order goes outside-in: block, then row, then column.

```python
# indexing a 1-D array
a1[0], a1[1], a1[-1]
```

`a1[-1]` is the last item. Negative numbers count backwards from the end, so
you never need `len(a1) - 1`.

```python
# indexing a 2-D array -> row first, then column
a2[1]          # whole row at index 1
```

One index on a 2-D array returns a whole **row**, not a single value. The row
is itself a 1-D array.

```python
a2[1][2], a2[1, 2]   # same element, two ways to write it
```

Both give `6`.

`a2[1][2]` does two steps: fetch row 1, then item 2 of that row.
`a2[1, 2]` does it in one step. The comma form is the NumPy way — it is faster
because it does not build the temporary row in between. Lists cannot do the
comma form at all.

```python
# indexing a 3-D array -> block first, then row, then column
a3[1]          # whole 2x2 block at index 1
```

```python
a3[1][0]       # row 0 inside block 1
```

```python
a3[1][0][1], a3[1, 0, 1]   # single element, two ways to write it
```

Each index peels off one layer: block → row → value.

```python
# indexing a 4-D array -> peel off one bracket layer at a time
a4[0]          # drops to 3-D: shape (1, 1, 4)
```

```python
a4[0][0][0], a4[0, 0, 0]      # drops to 1-D: array([4, 44, 444, 4444])
```

```python
a4[0, 0, 0, 2]      # single element -> 444
```

Four dimensions need four indexes to reach a value. Three indexes still leave
you holding an array.

### Negative indexing

```python
arr2
```

```python
arr2[-1]
```

```python
arr2[-2]
```

`-1` is the last item, `-2` is the second from last. Handy when you do not know
the length.

### When the index does not exist

Ask for an index past the end and NumPy raises `IndexError`. It does not return
`None` and it does not grow the array.

```python
a12 = np.array([1, 2, 3])
a12
```

```python
a12[3]      # only 0, 1, 2 exist -> IndexError
```

Read the message carefully: *"index 3 is out of bounds for axis 0 with size 3"*.
It tells you three things — the bad index, **which axis** it was on, and how big
that axis actually is. On a multi-dimensional array the axis number tells you
exactly which bracket position you got wrong.

```python
a4[0, 0, 1, 0]      # axis 2 has size 1, so index 1 is too big
```

Here the first two indexes were fine. Axis 2 has size 1, so its only valid
index is `0`.

## 7. Slicing — reading many values

The pattern is `array[start : stop : step]`.

- `start` is included
- `stop` is **excluded**
- `step` is how far to jump each time

Leave any of them out and NumPy fills in a sensible default: start `0`,
stop `end`, step `1`.

```python
arr2
```

```python
arr2[1:]        # from index 1 to the end
```

```python
arr2[:3]        # from the start up to (not including) index 3
```

`arr2[:3]` gives 3 items — indexes 0, 1, 2. Index 3 is left out. That
"stop is excluded" rule is the one people trip on.

```python
arr2[:]         # everything
```

```python
arr2[1:4:1]     # start 1, stop 4, step 1
```

```python
arr2[1:4:2]     # same window, but skip every other item
```

Step `2` takes index 1, skips 2, takes 3. Result: `[20, 40]`.

```python
arr2[:4:3]      # start 0, stop 4, step 3
```

Takes index 0, jumps 3, takes index 3. Result: `[10, 40]`.

```python
arr2[::4]       # whole array, step 4
```

```python
arr2[::]        # same as arr2[:]
```

`arr2[::4]` takes index 0 and index 4 → `[10, 50]`.

One important difference from lists: **a NumPy slice is a view, not a copy.**
Change the slice and you change the original array. A Python list slice gives
you a fresh copy. Use `.copy()` when you really want a separate array.

## 8. Looping through an array

A `for` loop works exactly like it does on a list.

```python
for i in arr2:
    print(i)
```

This is fine for printing. For maths, do **not** loop — write `arr2 * 2` and
NumPy does the whole array in one shot, in C. That is called vectorisation and
it is the main reason NumPy exists.

## 9. Searching with `np.where()`

`np.where(condition)` returns the **positions** where the condition is true —
not the values themselves.

```python
x = np.where(arr2 == 20)
print(x)
```

The result is `(array([1]),)` — a tuple holding one array. The tuple has one
entry per dimension, so a 1-D array gives a 1-element tuple. Read it as
"value 20 sits at index 1".

```python
x = np.where(arr2 == 50)
print(x)
```

Now a case with the same value repeated many times.

```python
arrr = np.array([1, 2, 3, 1, 2, 3, 1, 2, 3])
arrr
```

```python
x = np.where(arrr == 2)
print(x)
```

`(array([1, 4, 7]),)` — the value `2` appears at indexes 1, 4 and 7. Unlike
`list.index()`, which stops at the first hit, `np.where` finds every match.

## 10. Sorting

`np.sort()` returns a **new sorted array**. The original is untouched.

```python
arraysort = np.array([2, 1, 5, 3, 7, 5, 9, 8, 5, 7, 3])
arraysort
```

```python
np.sort(arraysort)
```

```python
arraysort      # unchanged
```

The sorted output came back, but `arraysort` is still in its original order.
`np.sort` is a copy-out function. To keep the result, assign it to a name.

## 11. Insert and delete

`np.insert(array, position, value)` puts a value at a position.

Same rule as sorting: it returns a **new array**. NumPy arrays are fixed size,
so "inserting" really means building a fresh, bigger array behind the scenes.

```python
np.insert(arr2, 5, 60)      # add 60 at the end position
```

```python
arr2      # still 5 items
```

The original never changed. Assign the result if you want to keep it.

```python
arrr1 = np.insert(arr2, 5, 60)
arrr1
```

```python
np.insert(arraysort, 0, 100)     # insert at the front
```

Position `0` pushes everything else one step to the right.

```python
np.delete(arraysort, 0)      # remove the item at index 0
```

`np.delete(array, position)` removes by **position**, not by value. To delete
by value, find the position with `np.where` first.

You can also delete several positions at once by passing a list.

```python
index = [0, 1, 2]
arrr2 = np.delete(arr2, index)
arrr2
```

```python
arr2      # again, untouched
```

Indexes 0, 1 and 2 were dropped, leaving `[40, 50]`.

## 12. Reversing with `np.flip()`

```python
np.flip(arr2)
```

```python
arr2
```

Reversed copy out, original unchanged — the same pattern yet again.

```python
arrflip = np.flip(arr2)
arrflip
```

## 13. Changing an array in place

Everything above returned a copy. These two actually modify the array.

First, plain assignment to an index.

```python
arr2[0] = 100
arr2
```

`arr2` really changed. Note the type rule still applies — assign `"hello"` into
an integer array and you get an error, because the array's dtype is locked in.

Second, the `.sort()` **method** (not the `np.sort()` function).

```python
arr2.sort()
```

```python
arr2
```

Two things to notice:

1. `arr2.sort()` printed nothing — it returns `None`, like `list.sort()`.
2. `arr2` itself is now sorted.

So: `np.sort(a)` gives a sorted copy, `a.sort()` sorts in place. Mixing them up
(`a = a.sort()`) is a classic bug — it sets `a` to `None`.

## 14. Appending

`np.append()` looks like `list.append()` but behaves completely differently.

```python
np.append(arr2, 1000)
```

```python
arr2      # not appended
```

The value was added in the returned array, but `arr2` is unchanged. Again — you
must assign the result.

And the method version simply does not exist.

```python
arr2.append(1000)
```

`AttributeError: 'numpy.ndarray' object has no attribute 'append'`.

This is not an oversight. A list can append cheaply because it keeps spare
room at the end. A NumPy array is one tight block of memory with no spare room,
so every "append" has to allocate a whole new block and copy everything across.
NumPy makes that cost visible by refusing to offer the method.

If you need to build up data item by item, use a Python list and convert once
at the end with `np.array(my_list)`.

## Recap

| What I wanted | What I wrote | Changes the original? |
|---|---|---|
| Make an array | `np.array([1, 2, 3])` | — |
| Count dimensions | `a.ndim` | no |
| Size of each dimension | `a.shape` | no |
| Force a type | `np.array([...], dtype=np.int32)` | — |
| Force a depth | `np.array([...], ndmin=4)` | — |
| Read one value | `a[1, 2]` | no |
| Read the last value | `a[-1]` | no |
| Read a range | `a[start:stop:step]` | no (but it is a view) |
| Find positions | `np.where(a == 20)` | no |
| Sorted copy | `np.sort(a)` | **no** |
| Sort in place | `a.sort()` | **yes** |
| Insert | `np.insert(a, pos, val)` | no |
| Delete | `np.delete(a, pos)` | no |
| Reverse | `np.flip(a)` | no |
| Add to the end | `np.append(a, val)` | no |
| Change one value | `a[0] = 100` | **yes** |

The pattern worth memorising: **almost every `np.something(array)` function
returns a copy.** If you forget to assign the result, your change disappears.
The only in-place operations here are direct assignment (`a[0] = 100`) and the
`.sort()` method.

### Three mistakes to avoid

1. One string in the data turns the **whole array** into strings — check
   `.dtype` when maths silently stops working.
2. `stop` in a slice is excluded. `a[:3]` gives 3 items, not 4.
3. `arr.append(x)` does not exist. Use `np.append(arr, x)` and assign the
   result — or build a Python list first.
