```python
from sklearn.datasets import load_breast_cancer
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
import matplotlib.pyplot as plt

data = load_breast_cancer()
data
```

```python
X = data.data
y = data.target

lda = LinearDiscriminantAnalysis(n_components=1)

X_lda = lda.fit_transform(X, y)
X_lda
```

```python
malignant_center = X_lda[y == 0].mean()
benign_center = X_lda[y == 1].mean()

print("Malignant Center:", malignant_center)
print("Benign Center:", benign_center)
```

```python
plt.figure(figsize=(8,4))

plt.scatter(X_lda[y==0],[0]*sum(y==0),label="Malignant")
plt.scatter(X_lda[y==1],[1]*sum(y==1),label="Benign")

plt.yticks([0,1],["Malignant","Benign"])
plt.xlabel("LDA Component")
plt.title("LDA Projection")
plt.legend()

plt.show()
```
